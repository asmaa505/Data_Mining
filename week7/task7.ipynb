```python
import pandas as pd
import numpy as np
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, fpgrowth, association_rules
```

```python
# 1. Create a synthetic dataset of 1000 transactions
# We have 15 unique items (Item1 to Item15)
np.random.seed(42)
items = [f'Item{i}' for i in range(1, 16)]

dataset = []
for _ in range(1000):
    # Each transaction contains a random number of items (between 1 and 8)
    num_items = np.random.randint(1, 9)
    transaction = np.random.choice(items, size=num_items, replace=False).tolist()
    dataset.append(transaction)

# Preprocessing: Convert dataset into a One-Hot Encoded DataFrame
te = TransactionEncoder()
te_ary = te.fit(dataset).transform(dataset)
df = pd.DataFrame(te_ary, columns=te.columns_)

# Set thresholds
min_support = 0.3
min_confidence = 0.6

print(f"--- Market Basket Analysis (Support: {min_support}, Confidence: {min_confidence}) ---")
```

```python
# 2. Apply Apriori Algorithm
print("\n[Running Apriori Algorithm...]")
frequent_itemsets_apriori = apriori(df, min_support=min_support, use_colnames=True)
rules_apriori = association_rules(frequent_itemsets_apriori, metric="confidence", min_threshold=min_confidence)

print(f"Number of rules found (Apriori): {len(rules_apriori)}")
print(rules_apriori[['antecedents', 'consequents', 'support', 'confidence', 'lift']].head())
```

```python
# 3. Apply FP-Growth Algorithm
print("\n[Running FP-Growth Algorithm...]")
frequent_itemsets_fp = fpgrowth(df, min_support=min_support, use_colnames=True)
rules_fp = association_rules(frequent_itemsets_fp, metric="confidence", min_threshold=min_confidence)

print(f"Number of rules found (FP-Growth): {len(rules_fp)}")
print(rules_fp[['antecedents', 'consequents', 'support', 'confidence', 'lift']].head())

# Comparison Summary
print("\n--- Summary ---")
if len(rules_apriori) == len(rules_fp):
    print("Both algorithms produced the same number of rules.")
else:
    print("Difference detected in rule generation.")
```
